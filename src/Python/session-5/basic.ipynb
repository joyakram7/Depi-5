```python
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def move(self):
        pass
class Bird (Animal):
    def move(self):
        print("move from Bird")
class Cat (Animal):
        def move(self):
            print("move from cat")
 
a =Bird()
a.move()
```

```python
class student:
   def __init__(self , name="joy" , age="5" ):
      self.name = name
      self.age = age
      self.grade = 0

   def std_info(self):
      return f"my name is {self.name} , my age is {self.age}"
```

```python
class model:
    def fit(self,x,y):
        raise NotImplementedError
    def predict(self,x):
        raise NotImplementedError
    
class LinearRegression(model):
    def fit (self,x,y):
        print("fitting linear regression")
    def predict(self, x):
        print("predicting with linear regression")
 

class DicisionTree(model):
    def fit(self, x, y):
        print("fitting Decision Tree ")
   
    def predict(self, x):
        print("predicting with Decision Tree ")

models = [LinearRegression(),DicisionTree()]
for model in models:
  model.fit (None , None)
  model.predict (None)
       
        
```
