```python
email = "Amit_ml@gmail.edu"
if email.count("@") == 1 and "." in email:
    print("Valid email")
else:
    print("Invalid email")
```

```python
username = email.split("@")[0]

print(username)
```

```python
domain = email.split("@")[1].split(".")[0]

print(domain)
```

```python
if email.endswith(".com"):
    print("Commercial Domain")

elif email.endswith(".edu"):
    print("Educational Domain")

else:
    print("Other Domain")
```
