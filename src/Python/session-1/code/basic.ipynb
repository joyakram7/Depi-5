```python
message = "###!!@mocleW EPGTQ!!!!6789"

clean_message = ""

for char in message:
    if char.isalpha() or char == " ":
        clean_message += char

print(clean_message)      
```

```python
words = clean_message.split()

print(words)
```

```python
first_word = words[0][::-1]
second_word = words[1]

first_word += second_word[0]
second_word = second_word[1:]

print(first_word.lower() +" "+second_word )
```
