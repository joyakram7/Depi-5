```python
def show_menu():
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Exit")


def get_operation():
    while True:
        show_menu()
        operation = input("Choose an operation: ")

        if operation in ["1", "2", "3", "4", "5"]:
            return operation
        else:
            print("Invalid choice. Please try again.")


def get_numbers():
    while True:
        try:
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))
            return num1, num2
        except ValueError:
            print("Please enter valid numbers.")


def calculate(num1, num2, operation):
    if operation == "1":
        return num1 + num2

    elif operation == "2":
        return num1 - num2

    elif operation == "3":
        return num1 * num2

    elif operation == "4":
        if num2 == 0:
            print("Cannot divide by zero.")
            return None
        return num1 / num2


def calculator():
    while True:
        operation = get_operation()

        if operation == "5":
            print("Thank you for using the calculator!")
            break

        num1, num2 = get_numbers()

        result = calculate(num1, num2, operation)

        if result is not None:
            print("The result is:", result)

        again = input("Do you want another calculation? (yes/no): ")

        if again.lower() == "no":
            print("Thank you for using the calculator!")
            break


calculator()
```

```python
""" 1"""
class MultiplicationTable:
    def __init__(self, number):
        self.number = number

    def print_table(self):
        for i in range(1, 11):
            print(f"{self.number} * {i} = {self.number * i}")




s1 = MultiplicationTable(9)

s1.print_table()
```

```python
"""2"""
class TwinPrime:

    def is_prime(self, number):
        if number < 2:
            return False

        for i in range(2, number):
            if number % i == 0:
                return False

        return True

    def find_twin_primes(self):
        for number in range(2, 100):
            if self.is_prime(number) and self.is_prime(number + 2):
                print(number, "and", number + 2)


J1 = TwinPrime()

J1.find_twin_primes()
```

```python
"""3"""
class PrimeFactors:

    def find_factors(self, number):
        self.number = number
        list = []

        for i in range(2, number + 1):
            while number % i == 0:
                list.append(i)
                number = number // i

        return list


s1 = PrimeFactors()

print(s1.find_factors(56))
```

```python
"""5"""
class PerfectNumbers :
    def __init__ (self,start , end):
        self.start= start
        self.end= end

    def is_perfect (self, number):
        self.number= number

        if number == 0:
            return True

        total = 0

        for i in range(1, number):
            if number % i == 0:
                total += i

        return total == number

    def find_perfect_numbers(self):
        perfect_numbers = []

        for number in range(self.start, self.end):
            if self.is_perfect(number):
                perfect_numbers.append(number)

        return perfect_numbers


s5 = PerfectNumbers(0, 100)

for number in s5.find_perfect_numbers():
    print(number)
```
